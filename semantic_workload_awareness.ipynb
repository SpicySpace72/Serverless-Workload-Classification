```python
import pandas as pd
import numpy as np
from scipy.stats import skew, kurtosis
from sklearn.linear_model import LinearRegression

print("Loading Phase 1 normalized dataset...")
# Load the matrix generated in Phase 1
df = pd.read_csv('preprocessed_full.csv', index_col='UID')
X = df.values

print("Extracting Phase 2 Mathematical Features. This will take a moment...")

# 1. Statistical Features (Time Domain)
mean_val = np.mean(X, axis=1)
var_val = np.var(X, axis=1)
skew_val = skew(X, axis=1)
kurt_val = kurtosis(X, axis=1)

# Custom Spike Ratio
epsilon = 0.01
max_val = np.max(X, axis=1)
r_spike = max_val / (mean_val + epsilon)

# 2. Frequency Features (Spectral Domain via FFT)
# Compute FFT and extract the maximum magnitude (excluding the DC component at index 0)
fft_mags = np.abs(np.fft.fft(X, axis=1))
dominant_fft = np.max(fft_mags[:, 1:], axis=1)

# 3. Trend Features (Linear Regression)
# Fit a simple linear model to extract the absolute slope for each time series
t = np.arange(X.shape[1]).reshape(-1, 1)
slopes = np.zeros(X.shape[0])
reg = LinearRegression()
for i in range(X.shape[0]):
    reg.fit(t, X[i])
    slopes[i] = np.abs(reg.coef_[0])

# Compile all engineered features into a tabular format for Snorkel/LightGBM
df_features = pd.DataFrame({
    'Mean': mean_val,
    'Variance': var_val,
    'Skewness': skew_val,
    'Kurtosis': kurt_val,
    'Spike_Ratio': r_spike,
    'Dominant_FFT_Mag': dominant_fft,
    'Abs_Trend_Slope': slopes
}, index=df.index)

output_path = 'phase2_features.csv'
df_features.to_csv(output_path)
print(f"Phase 2 Complete! Tabular feature matrix saved to: {output_path}")
```

```python
import pandas as pd
import numpy as np
import lightgbm as lgb
from sklearn.model_selection import train_test_split, StratifiedKFold, GridSearchCV
from sklearn.metrics import classification_report, confusion_matrix, f1_score, make_scorer
import matplotlib.pyplot as plt
import seaborn as sns

print("Loading Phase 3 Snorkel-labeled dataset...")
df = pd.read_csv('phase3_labeled_data.csv', index_col='UID')

feature_cols = ['Mean', 'Variance', 'Skewness', 'Kurtosis', 'Spike_Ratio',
                 'Dominant_FFT_Mag', 'Abs_Trend_Slope']
X = df[feature_cols]
y = df['Archetype_Label']

ALL_LABELS = [0, 1, 2, 3]
LABEL_NAMES = ['SPIKE', 'PERIODIC', 'RAMP', 'STATIONARY']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

print("Running GridSearchCV with Stratified K-Fold...")
param_grid = {
    'num_leaves': [15, 31, 63],
    'learning_rate': [0.01, 0.05, 0.1],
    'feature_fraction': [0.7, 0.8, 1.0],
}

macro_f1_all_classes = make_scorer(
    f1_score, average='macro', labels=ALL_LABELS, zero_division=0
)

skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
base_clf = lgb.LGBMClassifier(n_estimators=200, random_state=42, class_weight='balanced')

grid_search = GridSearchCV(
    estimator=base_clf,
    param_grid=param_grid,
    scoring=macro_f1_all_classes,
    cv=skf,
    n_jobs=-1,
)
grid_search.fit(X_train, y_train)

print(f"Best params: {grid_search.best_params_}")
print(f"Best CV macro F1 (all 4 classes counted): {grid_search.best_score_:.4f}")

clf = grid_search.best_estimator_
y_pred = clf.predict(X_test)

macro_f1 = f1_score(y_test, y_pred, average='macro', labels=ALL_LABELS, zero_division=0)
print("\n--- Pipeline Evaluation Metrics (corrected) ---")
print(f"Test Macro F1-Score = {macro_f1:.4f}\n")
print(classification_report(
    y_test, y_pred, labels=ALL_LABELS, target_names=LABEL_NAMES, zero_division=0
))

cm = confusion_matrix(y_test, y_pred, labels=ALL_LABELS)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=LABEL_NAMES, yticklabels=LABEL_NAMES)
plt.title('Confusion Matrix: Serverless Archetype Classification')
plt.ylabel('True Snorkel Label')
plt.xlabel('LightGBM Predicted Label')
plt.tight_layout()
plt.savefig('final_confusion_matrix.png', dpi=300)
print("Saved Confusion Matrix visual to: final_confusion_matrix.png")
print("Phase 4 Complete (GridSearch + Stratified K-Fold + corrected macro F1).")
```

```python
import pandas as pd
import numpy as np
from snorkel.labeling import labeling_function, PandasLFApplier, LFAnalysis
from snorkel.labeling.model import LabelModel

print("Loading Phase 2 feature matrix...")
df_features = pd.read_csv('phase2_features.csv', index_col='UID')

ABSTAIN = -1
SPIKE = 0
PERIODIC = 1
RAMP = 2
STATIONARY = 3

# --- Recalibrated thresholds, computed from your actual data instead of ---
# --- fixed guesses. This fixes the RAMP bug directly. ---
fft_90th = df_features['Dominant_FFT_Mag'].quantile(0.90)
ramp_thresh = df_features['Abs_Trend_Slope'].quantile(0.90)   # top 10% slope
stat_var_thresh = df_features['Variance'].quantile(0.10)      # bottom 10% variance

print(f"Recalibrated RAMP slope threshold (90th pct): {ramp_thresh:.6f}")
print(f"Recalibrated STATIONARY variance threshold (10th pct): {stat_var_thresh:.6f}")

@labeling_function()
def lf_detect_spike(x):
    return SPIKE if x.Spike_Ratio > 15.0 and x.Kurtosis > 10.0 else ABSTAIN

@labeling_function()
def lf_detect_periodic(x):
    return PERIODIC if x.Dominant_FFT_Mag > fft_90th and x.Variance > 0.1 else ABSTAIN

@labeling_function()
def lf_detect_ramp(x):
    return RAMP if x.Abs_Trend_Slope > ramp_thresh and x.Spike_Ratio < 5.0 else ABSTAIN

@labeling_function()
def lf_detect_stationary(x):
    return STATIONARY if x.Variance < stat_var_thresh else ABSTAIN

lfs = [lf_detect_spike, lf_detect_periodic, lf_detect_ramp, lf_detect_stationary]
applier = PandasLFApplier(lfs=lfs)
L_train = applier.apply(df_features)

# --- Diagnostic: see coverage/overlap/conflict PER LF before trusting the output ---
print("\n--- Labeling Function Diagnostic ---")
print(LFAnalysis(L=L_train, lfs=lfs).lf_summary())

print("\nTraining Snorkel LabelModel...")
label_model = LabelModel(cardinality=4, verbose=True)
label_model.fit(L_train=L_train, n_epochs=500, log_freq=100, seed=123)

y_train_preds = label_model.predict(L_train)
df_features['Archetype_Label'] = y_train_preds

print("\nLabel distribution after LabelModel resolution:")
print(pd.Series(y_train_preds).value_counts().sort_index())

df_labeled = df_features[df_features.Archetype_Label != ABSTAIN]
output_path = 'phase3_labeled_data.csv'
df_labeled.to_csv(output_path)
print(f"\nSaved: {output_path}")
print(f"Total labeled samples: {len(df_labeled)}")
```
